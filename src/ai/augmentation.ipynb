```python
import os
import cv2
import imgaug.augmenters as iaa
```

```python
os.chdir("..")
os.chdir("..")
path = "augmanted"
if not os.path.exists(path):
    os.mkdir(path)
```

```python
def augment_data(source_path, copies, new_path):
    img_list = []

    for img_name in os.listdir(source_path):
        img_path = os.path.join(source_path, img_name)
        img = cv2.imread(img_path)
        if img is not None:
            img_list.append(img)

    augmentation = iaa.Sequential([
        iaa.Rotate(rotate=(-15, 15)),
        iaa.Fliplr(0.5),
        iaa.Affine(
            translate_percent={"x": (-0.2, 0.2), "y": (-0.2, 0.2)},
            scale=(0.7, 1.2)
        ),
        iaa.LinearContrast((0.6, 1.5)),
    ])

    j = 0
    for i in range(copies):
        augmented_images = augmentation(images=img_list)
        for image in augmented_images:
            im_name = os.path.join(new_path, f"augmented_{j}.png")
            cv2.imwrite(im_name, image)
            j += 1
```

```python
source_path="dataset/TRAIN/2"
new_path="augmanted/TRAIN/2"
augment_data(source_path=source_path,copies=1,new_path=new_path)
```
